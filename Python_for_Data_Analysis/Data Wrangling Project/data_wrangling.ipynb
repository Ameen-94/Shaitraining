## 2- Handling Missing Values

```python
# How many null values are in each column
df.isnull().sum() 
```

### Dropping rows with no location information for the crashes

```python
# Looking for null values in location columns
df[
    (df["ON STREET NAME"].isnull() == True)
    & (df["OFF STREET NAME"].isnull() == True)
    & (df["CROSS STREET NAME"].isnull() == True)
    & (df["LOCATION"].isnull() == True)
    & (df["BOROUGH"].isnull() == True)
    & (df["ZIP CODE"].isnull() == True)
].count()
```

```python
# Taking indexes of rows with null values in location columns
indexes = df[
    (df["ON STREET NAME"].isnull() == True)
    & (df["OFF STREET NAME"].isnull() == True)
    & (df["CROSS STREET NAME"].isnull() == True)
    & (df["LOCATION"].isnull() == True)
    & (df["BOROUGH"].isnull() == True)
    & (df["ZIP CODE"].isnull() == True)
].index.tolist()
```

```python
# Dropping rows with null values in location columns
df.drop(index=indexes,axis=0, inplace=True)
```

```python
# After dropping rows with null values in location columns
df[
    (df["ON STREET NAME"].isnull() == True)
    & (df["OFF STREET NAME"].isnull() == True)
    & (df["CROSS STREET NAME"].isnull() == True)
    & (df["LOCATION"].isnull() == True)
    & (df["BOROUGH"].isnull() == True)
    & (df["ZIP CODE"].isnull() == True)
].count()
```

```python
# Dropping rows with unknown boroughs
df = df.dropna(subset=["BOROUGH"], ignore_index=True)
```

### Filling the rest of the missing values

```python
df["NUMBER OF PERSONS INJURED"] = df["NUMBER OF PERSONS INJURED"].fillna(0)
df["NUMBER OF PERSONS KILLED"] = df["NUMBER OF PERSONS KILLED"].fillna(0)
```

```python
df["CONTRIBUTING FACTOR VEHICLE 1"] = df["CONTRIBUTING FACTOR VEHICLE 1"].fillna(
    "Unknown"
)
df["CONTRIBUTING FACTOR VEHICLE 2"] = df["CONTRIBUTING FACTOR VEHICLE 2"].fillna(
    "Unknown"
)
df["CONTRIBUTING FACTOR VEHICLE 3"] = df["CONTRIBUTING FACTOR VEHICLE 3"].fillna(
    "Unknown"
)
df["CONTRIBUTING FACTOR VEHICLE 4"] = df["CONTRIBUTING FACTOR VEHICLE 4"].fillna(
    "Unknown"
)
df["CONTRIBUTING FACTOR VEHICLE 5"] = df["CONTRIBUTING FACTOR VEHICLE 5"].fillna(
    "Unknown"
)
```

```python
df["LOCATION"] = df["LOCATION"].fillna("Unknown")
df["ON STREET NAME"] = df["ON STREET NAME"].fillna("Unknown")
df["OFF STREET NAME"] = df["OFF STREET NAME"].fillna("Unknown")
df["CROSS STREET NAME"] = df["CROSS STREET NAME"].fillna("Unknown")
df["ZIP CODE"] = df["ZIP CODE"].fillna("Unknown")
df["LONGITUDE"] = df["LONGITUDE"].fillna(0)
df["LATITUDE"] = df["LATITUDE"].fillna(0)
```

```python
df["VEHICLE TYPE CODE 1"] = df["VEHICLE TYPE CODE 1"].fillna("Unknown")
df["VEHICLE TYPE CODE 2"] = df["VEHICLE TYPE CODE 2"].fillna("Unknown")
df["VEHICLE TYPE CODE 3"] = df["VEHICLE TYPE CODE 3"].fillna("Unknown")
df["VEHICLE TYPE CODE 4"] = df["VEHICLE TYPE CODE 4"].fillna("Unknown")
df["VEHICLE TYPE CODE 5"] = df["VEHICLE TYPE CODE 5"].fillna("Unknown")
```

## 5- Filtering and Grouping Data

About 300 thousands crashes out of 1.5 million happened in Manhattan

```python
# People injured in crashes based of the borough it happened in
df.groupby("BOROUGH")["NUMBER OF PERSONS INJURED"].sum()
```

```python
# People killed in crashes based of the borough it happened in
df.groupby("BOROUGH")["NUMBER OF PERSONS KILLED"].sum()
```

```python
# number of crashes for every day
df["CRASH YEAR"] = df["CRASH DATE"].dt.year
df.groupby("CRASH YEAR")["CRASH YEAR"].size()
```

```python
# A pivot table showing the number of people injured in different boroughs every year
df.pivot_table(
    values="NUMBER OF PERSONS INJURED",
    index="BOROUGH",
    columns="CRASH YEAR",
    aggfunc=sum,
)
```

```python
# A correlation matrix showing the correlation between people and pedestrians killed and injured
df_corr = df[["NUMBER OF PERSONS INJURED","NUMBER OF PERSONS KILLED","NUMBER OF PEDESTRIANS INJURED","NUMBER OF PEDESTRIANS KILLED"]]
df_corr.corr(numeric_only=True)
```

## 6- Statistical Summaries

```python
print(f"The average rate of people injured in crashes is: {round(df["NUMBER OF PERSONS INJURED"].mean(),3)}")
print(f"The average rate of people killed in crashes is: {round(df["NUMBER OF PERSONS KILLED"].mean(),3)}")
```

```python
print(f"The total number of people injured in crashes is: {df["NUMBER OF PERSONS INJURED"].sum()}")
print(f"The total number of people killed in crashes is: {df["NUMBER OF PERSONS KILLED"].sum()}")
```

```python
print(f"The maximum number of people injured in a single crash is: {df["NUMBER OF PERSONS INJURED"].max()}")
print(f"The maximum number of people killed in a single crash is: {df["NUMBER OF PERSONS KILLED"].max()}")
```

```python
max_borough = df["BOROUGH"].value_counts().idxmax()
print(f"The borough with the most crashes is {max_borough}")
```

```python
num_of_injuries = df[df["NUMBER OF PERSONS INJURED"] > 0].value_counts().count()
percentage = (num_of_injuries/df.shape[0]) * 100
print(f"The percentage of injured people in crashes is {round(percentage,2)}%")
```
