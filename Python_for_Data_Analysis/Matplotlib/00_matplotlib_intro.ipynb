
# Introduction to Matplotlib

Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python.
It is one of the most widely used data visualization libraries in the Python data science ecosystem,
and it provides a wide variety of plot types and customization options.
Whether you are creating line graphs, bar graphs, histograms, or more complex visualizations,
Matplotlib provides the tools you need to generate high-quality figures.



## Line Plots

Line plots are used to display data points connected by straight line segments.
They are commonly used to visualize data over a continuous interval or time span.


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create line plot
plt.plot(x, y)
plt.title('Line Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
plt.plot(x, y, lw=30)
```


## Scatter Plots

Scatter plots display individual data points on a two-dimensional graph,
and are useful for showing the relationship between two variables.


```python
import numpy as np
import matplotlib.pyplot as plt
# Sample data
# x = sorted(np.random.rand(50))
x = np.random.rand(50)
# y = np.random.rand(50)
y = sorted(np.random.rand(50))
s = np.linspace(50, 250, 50)

my_color = [tuple((0.54,0.17,0.89,float(alpha))) for alpha in np.linspace(0.5, 1, 50)]

# Create scatter plot
plt.scatter(x, y, s=s, color=my_color, marker="^")
plt.title('Scatter Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
x= np.random.rand(50)
y= np.random.rand(50)
plt.scatter(x,y,s=900)
plt.show
```


## Histograms

Histograms are used to represent the distribution of a dataset.
It divides the data into intervals (bins), and the height of the bars represents the number of data points in each bin.


```python
np.linspace(0.5, 1, 5)
```

```python
sorted(np.random.randn(100)) 
```

```python
# Sample data
datasets = [np.random.randint(100) for _ in range (5)]
# datasets = [np.random.randint(0,1000,1000)]

my_color = [(0.54,0.17,0.89, 0.5), (0.4,0,0.04, 0.625), (0.4,0.68,0.08, 0.75), (0.4,0.45,0.13, 0.875), (0.4,0.45,0.36)]

plt.hist(datasets, bins=15, color=my_color[2])
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend([1,2,3,4,5])
plt.grid(True)
plt.show()
```

```python

# Sample data
data = np.random.randn(1000) #n

# Create histogram
plt.hist(data, bins=30, edgecolor='black')
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
datasets = [np.random.randn(1000) for _ in range(5)]
c = [(0.54, 0.17, 0.89, float(alpha)) for alpha in np.linspace(0.5, 1, 5)]

plt.hist(datasets, bins=30, edgecolor='black', color=c)
plt.show()
```


## Box Plots

Box plots (also known as box-and-whisker plots) are used to display the distribution of a dataset based on a five number summary:
minimum, first quartile, median, third quartile, and maximum.


```python

# Sample data
data = [np.random.randint(0,250,1000), np.random.randint(250,500,1000), np.random.randint(500,750,1000), np.random.randint(750,1000,1000)]

# Create box plot
plt.boxplot(data)
plt.title('Box Plot Example')
plt.xlabel('Category')
plt.ylabel('Value')
# plt.grid(True)
plt.show()
```


## Pie Charts

Pie charts are circular statistical graphics divided into slices to illustrate numerical proportion.


```python

# Sample data
labels = ['A', 'B', 'C', 'D', 'E']
sizes = [10, 30, 40, 10, 10]

my_color = [(0.54,0.17,0.89, 0.5), (0.4,0,0.04, 0.625), (0.4,0.68,0.08, 0.75), (0.4,0.45,0.13, 0.875), (0.4,0.45,0.36,0.7)]

# Create pie chart
plt.pie(sizes, labels=labels, autopct='%1.f%%', startangle=180, colors=my_color)  #Try without startangle
plt.title('Pie Chart Example')
plt.show()
```


## Bar Charts

Bar charts are used to display quantities corresponding to different categories.
They consist of rectangular bars, where the length of the bar represents the quantity of the category.


```python

# Sample data
categories = ['A', 'B', 'C', 'D']
values = [np.random.randint(0,100), np.random.randint(0,100),np.random.randint(0,100),  np.random.randint(0,100)]

# Create bar chart
plt.bar(categories, values)
plt.title('Bar Chart Example')
plt.xlabel('Category')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```
