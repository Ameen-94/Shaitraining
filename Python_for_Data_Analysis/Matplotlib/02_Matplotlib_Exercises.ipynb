## Exercise 1

** Follow along with these steps: **
* ** Create a figure object called fig using plt.figure() **
* ** Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax. **
* ** Plot (x,y) on that axes and set the labels and titles to match the plot below:**

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
x = np.linspace(0,200)
y = np.linspace(0,200)
ax.plot(x,y,'blue')
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_title("X VS Y Plot")
```

## Exercise 2
** Create a figure object and put two axes on it, ax1 and ax2. Located at [0,0,1,1] and [0.2,0.5,.2,.2] respectively.**

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])
ax2 = fig.add_axes([0.2,0.5,0.2,0.2])
ax1.set_title("Main Plot (ax1)")
ax2.set_title("Inset Plot (ax2)")
ax1.set_xlabel("X for ax1")
ax1.set_ylabel("Y for ax1")
ax2.set_xlabel("X for ax2")
ax2.set_ylabel("Y for ax2")
```

** Now plot (x,y) on both axes. And call your figure object to show it.**

```python
ax1.plot(x,y,'blue')
ax2.plot(x,y,'red')
fig
```

## Exercise 3

** Create the plot below by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,.4,.4]**

```python
fig = plt.figure()
ax1 = fig.add_axes([0,0,1,1])
ax2 = fig.add_axes([0.2,0.5,0.4,0.4])
ax1.set_title("Main Plot (ax1)")
ax2.set_title("Inset Plot (ax2)")
ax1.set_xlabel("X for ax1")
ax1.set_ylabel("Y for ax1")
ax2.set_xlabel("X for ax2")
ax2.set_ylabel("Y for ax2")
```

** Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot:**

```python
x = np.linspace(0,100)
y = x*2
z = x**2
ax1.plot(x,z,'blue')
ax2.plot(x,y,'blue')
ax2.set_xlim([20,22])
ax2.set_ylim([30,50])
ax1.set_title("")
ax2.set_title("Inset Plot")
ax1.set_xlabel("X")
ax1.set_ylabel("Z")
ax2.set_xlabel("X")
ax2.set_ylabel("Y")
fig
```

## Exercise 4

** Use plt.subplots(nrows=1, ncols=2) to create the plot below.**

```python
plt.subplots(nrows=1,ncols=2)
```

** Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style**

```python
fig,axes = plt.subplots(nrows=1,ncols=2)
axes[0].plot(x, y, color = 'blue', ls = '--', lw = 3)
axes[0].set_title("X VS Y Plot (Dashed)")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[1].plot(x,z,color = 'red', ls = ':')
axes[1].set_title("X VS Z Plot (Dotted)")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Z")
plt.tight_layout()
```

** See if you can resize the plot by adding the figsize() argument in plt.subplots() are copying and pasting your previous code.**

```python
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,10))
axes[0].plot(x, y, color = 'blue', ls = '--', lw = 3)
axes[0].set_title("X VS Y Plot (Dashed)")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[1].plot(x,z,color = 'red', ls = ':')
axes[1].set_title("X VS Z Plot (Dotted)")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Z")
plt.tight_layout()
```
