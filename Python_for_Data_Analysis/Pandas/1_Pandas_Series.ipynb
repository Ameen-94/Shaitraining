# Introduction to Pandas Library

Pandas is a powerful and widely-used Python library designed for **data manipulation and analysis**.

It provides easy-to-use **data structures** and functions that make working with structured data (like tables) more efficient and intuitive.

In pandas, **Series and DataFrames** are two fundamental data structures that serve different purposes

# Introduction to Series

*   `Pandas Series`: A Series is a one-dimensional labeled array that can hold any data type (integers, strings, floats). It is similar to a column in an Excel spreadsheet or a single column in a DataFrame.

* A Series is very similar to a NumPy array (in fact it is built on top of the NumPy array object).

* What differentiates the NumPy array from a Series, is that a Series can have axis labels, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data, it can hold any arbitrary Python Object.

# Creating a Series

````python
# Create a Series from a list --automatic indexing

data = [10, 20, 30, 40]
series = pd.Series(data)
print(series)
````

````python
series[3]
````

````python
#Add custom indexing

data = [10, 20, 30, 40]
series = pd.Series(data = data, index = ['a', 'b', 'c', 'd'])
series
````

````python
#Add a name for the series

series.name = 'numbers and letters'
series
````



---



````python
# Create a Series from a dictionary

data = {'a': 10, 'b': 20, 'c': 30}
series = pd.Series(data)
series
# the dictionary keys become the index of the Series.
````



---

````python
# Create a Series from a scalar value

series = pd.Series(189, index=['a', 'b', 'c', 'd'])
series
````

````python
from uuid import uuid4
series = pd.Series(['a', 'b', 'c', 'd', 'e'], index=[str(uuid4())[:4] for _ in range(5)])
series
````



---



````python
# Create a Series from array

arr = np.array([10, 20, 30])
pd.Series(arr)
````

````python
labels=["First", "Second", "Third"]
pd.Series(arr, labels)
````

Create a Series from a dictionary that results in the following series:

Bananas         0.75

Cherries        2.50

Name the Series "Fruit Prices"

# Convert a series to dictionary

````python
# You can convert a series to dictionary  --> use to_dict

temperatures = pd.Series([23, 20, 29, 30, 20, 22, 27], index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
temperatures_dict = temperatures.to_dict()

print(temperatures_dict)
````

````python
temperatures.to_csv("temperatures.csv")
````

````python
# calculate the average temperature
average_temp = temperatures.mean()
print(f"Average Temperature: {average_temp}")
````

Create a Pandas Series called scores using the following student exam scores:

Print the original Series.

*Bonus (Optional): Count how many students passed.*

# Using an Index

The key to using a Series is understanding its index. Pandas makes use of these index names or numbers by allowing for fast look ups of information (works like a hash table or dictionary).

Let's see some examples of how to grab information from a Series. Let us create two sereis, ser1 and ser2:

````python
ser1 = pd.Series([1, 2, 3, 4], index = ['USA', 'Germany','Russia', 'Japan'])
ser1
````

````python
ser2 = pd.Series([1,2,5], index = ['USA', 'Germany','Italy'])
ser2
````

Operations are then also done based off of index:

````python
ser1 + ser2
````

````python
my_results = ser1 + ser2
my_results.dropna()  
````
