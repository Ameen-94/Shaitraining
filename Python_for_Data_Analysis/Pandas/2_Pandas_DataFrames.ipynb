# Introduction to DataFrames

*  `Pandas DataFrame`: A DataFrame is a two-dimensional labeled data structure with columns of potentially different data types. It can be thought of as a collection of Series objects, where each Series represents a column in the DataFrame.

# Creating a DataFrames

```python
#If we were to use a series for this data, this is how it would look like:

data = {'Name': ['Omar', 'Mohammad', 'Ross'],
        'Age': [25, 30, 35],
        'Salary': [50000, 60000, 70000]}

df = pd.Series(data)
df
```

```python
# However, using a DataFrame it would look like this:

data = {'Name': ['Omar', 'Mohammad', 'Ross'],
        'Age': [25, 30, 35],
        'Salary': [50000, 60000, 70000]
        }

df = pd.DataFrame(data)
df
```

# Aggregation Functions

```python
data = [{'Name': 'Ayman', 'Age': 28, 'Salary': 55000}, # it represents the first row
        {'Name': 'Omar', 'Age': 30, 'Salary': 80000},# it represents the second row
        {'Name': 'Mohammad', 'Age': 30, 'Salary': 60000} # it represents the third row
        ]
df = pd.DataFrame(data)

df
```

```python
df['Age'].mean()
```

```python
mean_age = df['Age'].mean().round(1)
print(f"Mean Age: {mean_age}")

mean_age = int(df['Age'].mean().round(0))
print(f"Mean Age: {mean_age}")

mean_age = round(df['Age'].mean(), 1)
print(f"Mean Age: {mean_age}")
```

```python
# Sum of all salaries
total_salary = df['Salary'].sum()
print(f"Total Salary: {total_salary}")
```
