# Filtering the data

```python
data = {
    'Name': ['Omar', 'Mohammad', 'Tuqa', 'Zaina', 'Malik' ],
    'Age': [25, 30, 35, 20, 45],
    'City': ['Amman', 'Damascus', 'Damascus', 'Yafa', 'Irbid'],
    'Score': [85.5, 90.3, 78.9, 92.5, 88.0]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df['Age'] > 30
```

```python
df[df['Age'] > 30]
```

```python
df["City"] == "Yafa"
```

```python
df[df["City"]=="Yafa"]
```

```python
df[(df["City"]=="Damascus") | (df["Age"]>35)]
```

```python
df[(df["City"]=="Damascus") & (df["Age"]>30)]
```

```python
df[df["City"]=="Zarqa"]
```

```python
df[(df['Name'] == 'Mohammad') | (df['Name'] == 'Omar')  | (df['Name'] == 'Tuqa')]
```

```python
# using isin
names = ['Tuqa' , 'Omar' , 'Zaina', 'batool']

multiple_values = df[ df['Name'].isin(names)  ]
multiple_values
```

Task 2:

Write a pandas code to filter the data based on two conditions price is greater than or equal 800
and   quantity is greater than or equal 5

```python
data = [
    {'Product': 'Laptop', 'Price': 1200, 'Quantity': 5, 'Discount': 10},
    {'Product': 'Tablet', 'Price': 600, 'Quantity': 10, 'Discount': 15},
    {'Product': 'Smartphone', 'Price': 800, 'Quantity': 7, 'Discount': 5},
    {'Product': 'Monitor', 'Price': 300, 'Quantity': 12, 'Discount': 20}
]

products_df = pd.DataFrame(data)
products_df
```

```python
products_df[(products_df["Price"] >= 800) & (products_df["Quantity"] <= 5)] 
```
