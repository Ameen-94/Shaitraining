# Add column by conditions

```python
df[df['Age'] > 30]
```

```python
# using np.where to add a column based on a condition
df['Senior'] = np.where(df['Age'] > 30, 'Yes', 'No')
df
```

# Add column by multiple conditions

```python
# Multiple Conditions

def age_category(age):

    if age >= 20 and age <= 25:
        return  'Yes'
    elif age >= 30 and age <= 34:
        return 'Maybe'
    else:
        return 'No'

df['Senior'] = df['Age'].apply(age_category)
df
```

# Quick Reviews

```python
data = [
    {'Product': 'Laptop', 'Price': 1200, 'Quantity': 5, 'Discount': 10},
    {'Product': 'Tablet', 'Price': 600, 'Quantity': 10, 'Discount': 15},
    {'Product': 'Smartphone', 'Price': 800, 'Quantity': 7, 'Discount': 5},
    {'Product': 'Monitor', 'Price': 300, 'Quantity': 12, 'Discount': 20}
]

df = pd.DataFrame(data)

df
```

```python
# use  np.where to add a new column high_price that marks products as "Yes" if the
# Price is greater than 700, and "No" otherwise.
df['High_Price'] = np.where(df['Price'] > 700 ,'Yes' , 'No')
df
```

```python
# define a function that  categorizes products named category based on Quantity, 'high'
# if quantity is greater than 10 , 'medium' if quantity > 5  else : 'low stock'

def categorizes(Quantity):
  if Quantity >= 10:
        return  'high'
  elif Quantity >= 5:
        return 'medium'
  else:
        return 'low'

df['Stock'] = df['Quantity'].apply(categorizes)
df
```

```python
data = pd.DataFrame({
    'A': [1, 6, 3, 4, 5],
    'B': [5, 4, 3, 5, 1]
})

data
```

```python
# using np.where with two conditions from different columns
data['Result'] = np.where((data['A'] > 2) & (data['B'] < 4), 'Condition Met', 'Condition Not Met')

data
```
