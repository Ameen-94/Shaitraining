# Missing Data

Let's show a few convenient methods to deal with Missing Data in pandas:

````python
import numpy as np
import pandas as pd
````

````python
df = pd.DataFrame({'A':[1,2,np.nan, 4, np.nan, 7, 3],
                  'B':[5,np.nan,np.nan, 8, 9, np.nan, 10],
                  'C':[1,2,3, 4, 5, 6, 7]})
````

````python
df
````

````python
df.isna().sum()
````

````python
df.dropna()
````

````python
df.dropna(axis=1)
````

````python
df.dropna(thresh=2)  #Keep any row that has at least 2 non-NaN values
````

````python
df.fillna(value='Missing')
````

````python
df['A'].fillna(value=df['A'].mean())
````

````python
df.fillna(method='ffill', limit=2) # what if limit was 2? (2 consecutive NaNs will be filled)
````

````python
df.fillna(method='bfill', limit=1)
````

````python
df # for comparison
````

````python
df.interpolate()
````

# Task 1:

Using the following DataFrame:
```
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales_A': [1000, np.nan, 1200, np.nan, 1600, 1700],
    'Sales_B': [900, 950, np.nan, np.nan, 1100, 1150],
    'Remarks': ['OK', np.nan, 'OK', 'Late', np.nan, 'OK']
}
```
perform the following operations:


1.  Drop all rows that have all values missing. (you can use a quick Google Search in here!)
2.  Drop all rows that have less than 2 non-missing values.
3.  Fill missing values in Remarks with 'No Info'.
4.  Fill missing values in Sales_A with the mean of that column.
5.  Use forward fill to fill only 1 missing value in Sales_B.
6.  Use interpolation to fill remaining missing values in Sales_B.



````python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',np.nan],
    'Sales_A': [1000, np.nan, 1200, np.nan, 1600, 1700,np.nan],
    'Sales_B': [900, 950, np.nan, np.nan, 1100, 1150,np.nan],
    'Remarks': ['OK', np.nan, 'OK', 'Late', np.nan, 'OK',np.nan]
}

df = pd.DataFrame(data)
df
````

````python
# Step 1: Drop rows where all values are missing (just in case)
df = df.dropna(how='all')
df
````

````python
# Step 2: Drop rows with fewer than 2 non-null values
df = df.dropna(thresh=2)
df
````

````python
# Step 3: Fill missing values in 'Remarks' with 'No Info'
df['Remarks'] = df['Remarks'].fillna('No Info')
df
````

````python
# Step 4: Fill missing values in 'Sales_A' with the column's mean
df['Sales_A'] = df['Sales_A'].fillna(df['Sales_A'].mean())
df
````

````python
# Step 5: Forward fill one missing value only in 'Sales_B'
df['Sales_B'] = df['Sales_B'].fillna(method='ffill', limit=1)
df
````

````python
# Step 6: Interpolate remaining missing values in 'Sales_B'
df['Sales_B'] = df['Sales_B'].interpolate()
df
````
