# SORTING

````python
import pandas as pd
import numpy as np
````

````python
data = {
    'Name': ['Yafa', 'Karam', 'Emad', 'Zain'],
    'Age': [22, 30, 35, 22],
    'Salary': [60000,50000, 70000, 55000]
}

df = pd.DataFrame(data)

df
````

````python
# Sort by 'Age'---> by default it sorts in ascending order (تصاعدي)
df_sorted = df.sort_values(by='Age')
df_sorted
````

````python
# Sort by Age and then by Salary (if two people have the same age, this sorts them by their salary)
df_sorted_multi = df.sort_values(by=['Age', 'Salary'])
df_sorted_multi
````

````python
# sort by salary in descending order

df_sorted_desc = df.sort_values(by='Salary', ascending=False)
df_sorted_desc
````

````python
df_sorted_desc = df.sort_values(by=['Salary', "Age"], ascending=False) # Why is Karam's record last?
df_sorted_desc
````

````python
# real world Examples

"""
sorting the sales data helps identify the most popular products,
which can help make inventory decisions,
example : if headphones are the top-selling product, you might want to buy more of them
"""
sales_data = {
    'Product': ['Laptop', 'Headphones', 'Keyboard', 'Mouse'],
    'Units Sold': [150, 300, 200, 250],
    'Date': pd.to_datetime(['2024-01-15', '2024-01-10', '2024-01-12', '2024-01-14'])
}

sales_df = pd.DataFrame(sales_data)
````

# Task 3:
Given the following data:
```
data = {
    'Name': ['Salma', 'Ali', 'Nada', 'Tariq', 'Lina'],
    'Age': ['25', '32', None, '27', '29'],
    'Satisfaction': ['4.5', '3.8', None, '5.0', '3.0']
}
```
Perform the following:
1. Display the DataFrame to see what you’re working with.

2. Convert 'Age' and 'Satisfaction' to float using .astype(float).

3. Fill missing Age values with the mean Age.

4. Fill missing values in the 'Satisfaction' column with a fixed value: 3.0.

5. Print the cleaned DataFrame with correct types and filled values.

````python
data = {
    'Name': ['Salma', 'Ali', 'Nada', 'Tariq', 'Lina'],
    'Age': ['25', '32', None, '27', '29'],
    'Satisfaction': ['4.5', '3.8', None, '5.0', '3.0']
}

df = pd.DataFrame(data)

# Before cleaning
print("Before cleaning:")
print(df)
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isna().sum())
````

````python
# Step 2: Convert to numeric
# Convert to float
df['Age'] = df['Age'].astype(float)
df['Satisfaction'] = df['Satisfaction'].astype(float)
print(df.dtypes)
df
````

````python
# Step 3: Fill missing Age with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
````

````python
# Step 4: Fill Satisfaction with 3.0
df['Satisfaction'] = df['Satisfaction'].fillna(3.0)
df
````

````python
# Final cleaned DataFrame
print("\nAfter cleaning:")
print(df)
print("\nData types:\n", df.dtypes)
````
