# groupby in Pandas









**The groupby operation in pandas is a powerful tool used to split your data into groups based on some criteria, apply a function to each group, and combine the results. It's similar to the split-apply-combine strategy:**

*   Split: The data is split into groups based on some column(s).
*   Apply: A function is applied to each group independently.


*   Combine: The results are combined back into a DataFrame or Series.

**Why Use groupby?**



*   It helps you explore and analyze data by grouping similar items.

*   You can calculate summary statistics (like mean, sum, count, etc.) for groups of data.


*   It is especially useful for handling time series, categorical data, or datasets where you need to aggregate by a specific feature.

````python
df['job_title']
````

````python
"""
Grouping by a Single Column
Problem: Find the average salary for each job title.
"""

# option 1

average_salary_by_job = df.groupby('job_title')['salary_in_usd'].mean()
average_salary_by_job
````

````python
average_salary_by_job.count()
````

````python
# option 2
average_salary_by_job = df.groupby('job_title').agg({"salary_in_usd":"mean"})
average_salary_by_job
````

````python
a = average_salary_by_job.count()
print(a)
````

````python
# Find the total salary paid for each combination of experience_level and employment_type.

total_salary_by_experience_employment = df.groupby(['experience_level', 'employment_type'])['salary_in_usd'].sum()
total_salary_by_experience_employment
````

# Task : find the total and average salary for each job_title

You can perform multiple operations on different columns within the same group using the agg() function like what we have seen above


````python
# approach one

salary_stats_by_job = df.groupby('job_title').agg({"salary_in_usd":['mean','median']})
salary_stats_by_job.reset_index()
````

````python
# approach two (recommended)
salary_stats_by_job = df.groupby('job_title').agg(
    total_salary=('salary', 'sum'),
    average_salary=('salary', 'mean'),
)

salary_stats_by_job
````

````python
# find the total and average for (salary and salary_in_usd) for each job_title and experience level  using the recommended way
y=df.groupby(['job_title','experience_level']).agg(
    total_salary=('salary','sum'),
    average_salary=('salary','mean'),
    total_salary_in_usd=('salary_in_usd','sum'),
    average_salary_in_usd=('salary_in_usd','mean')

)
y
````

````python
# Count the number of employees in each country (employee_residence)
employee_count_by_residence = df.groupby('employee_residence').size()
employee_count_by_residence
````

1. Create a pivot table that shows the total units sold for each Product in each Region.

* Use Region as the rows (index)

* Use Product as the columns (columns)

* Aggregate using the sum of Units_Sold

2. create another pivot table showing the average units sold by each Salesperson for each Product.

Bonus: Replace any NaN values with 0 in the pivot tables.


````python
data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East', 'West', 'West', 'North', 'South'],
    'Product': ['Apples', 'Oranges', 'Apples', 'Bananas', 'Apples', 'Bananas', 'Oranges', 'Bananas', 'Apples', 'Oranges'],
    'Salesperson': ['Ali', 'Lina', 'Sara', 'Khalid', 'Lina', 'Ali', 'Sara', 'Khalid', 'Sara', 'Ali'],
    'Units_Sold': [50, 40, 30, 25, 20, 35, 60, 45, 40, 20]
}

df = pd.DataFrame(data)
df
````

````python
pivot1 = df.pivot_table(
    values='Units_Sold',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

pivot1
````

````python
pivot1.fillna(0)
````
