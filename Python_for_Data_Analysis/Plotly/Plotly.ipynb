```python
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['week_of_year'] = df['date'].dt.isocalendar().week
df['week_start_dt'] = df['date'] - pd.to_timedelta(df['date'].dt.weekday, unit='D')
df
```

```python
df.groupby(['week_start_dt']).agg({"price_per_case":'sum','Quantity':'sum'}).reset_index()
```

```python
df_grouped = df.groupby(['week_start_dt' , 'size','Company','week_of_year',
            'year','Company_Region']).agg({"price_per_case":'sum','Quantity':'sum'}).reset_index()

df_grouped
```

```python
df_grouped = df_grouped[df_grouped['Company'] == 121]
```

```python
df_grouped
```

```python
df_grouped['size'].unique().tolist()
```

```python


# fig = make_subplots(rows=1, cols=1, shared_axes=True)
fig = go.Figure()

# List of sizes
sizes = df_grouped['size'].unique().tolist()

for size in sizes:

    filtered_data = df_grouped[(df_grouped['size'] == size) & (df_grouped['year'] == 2021)].sort_values('week_start_dt')

    fig.add_trace(go.Scatter(
        x=filtered_data['week_start_dt'],
        y=filtered_data['Quantity'],
        mode='lines+markers',  # Use both lines and markers for better visibility
        name=f'quantity Size {size}'
    ))

fig.update_layout(
    title='Quantity Trends by Size Over Time (2021) for commany 121',
    xaxis_title='Date',
    yaxis_title='Qunatity Per Case',
    # width=1000,
    # height=400,
)


fig.show()
```

```python
fig = go.Figure()

sizes = df_grouped['size'].unique()
years = df_grouped['year'].unique()

for size in sizes:
    for year in years:
      filtered_data = df_grouped[(df_grouped['size'] == size) & (df_grouped['year'] == year) & (df_grouped['week_of_year']<=50)].sort_values('week_start_dt')
      fig.add_trace(go.Scatter(
          x=filtered_data['week_of_year'],
          y=filtered_data['Quantity'],
          mode='lines+markers',  # Use both lines and markers for better visibility
          name=f'quantity {size} year {year}'
      ))

# Update layout to improve aesthetics
fig.update_layout(
    title='Quantity Trends by Size Over Time (2021) for commany 121',
    xaxis_title='Date',
    yaxis_title='Qunatity Per Case',
    # width=1000,
    # height=400,
)


# Show the updated figure
fig.show()
```

```python
# last but not least

dates = pd.date_range(start='2024-08-01', periods=8, freq='D')
actual_values = np.random.randint(20, 40, size=8)
predictions = actual_values + np.random.randint(-10, 10, size=8)

data = {'Date': dates,
    'Actual': actual_values,'Prediction': predictions}
df = pd.DataFrame(data)

df
```

```python
df['difference'] = abs(df['Actual'] - df['Prediction'])
```

```python
# how to add difference line ?

fig = go.Figure()

fig.add_trace(go.Scatter(x=df['Date'], y=df['Actual'],
                        mode='lines+markers',name=f'Actual'))

fig.add_trace(go.Scatter(x=df['Date'], y=df['Prediction'],
                        mode='lines+markers', name=f'prediction'))

fig.add_trace(go.Scatter(x=df['Date'], y=df['difference'],
                        mode='lines+markers', name=f'difference'))

fig.update_xaxes(title_text='Date', tickformat='%Y-%m-%d',
                tickvals=df['Date'], tickmode='array')

fig.update_layout(title='Actual vs Prediction')

fig.show()
```
