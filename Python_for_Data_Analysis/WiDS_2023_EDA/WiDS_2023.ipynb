## 2- Cleaning The Data

```python
null_dict = df.isnull().sum().to_dict()
null_dict
```

```python
# interpolate missing values for the following columns
df["nmme0-tmp2m-34w__ccsm30"].interpolate(inplace=True)
df["nmme-tmp2m-56w__ccsm3"].interpolate(inplace=True)
df["nmme-prate-34w__ccsm3"].interpolate(inplace=True)
df["nmme0-prate-56w__ccsm30"].interpolate(inplace=True)
df["nmme0-prate-34w__ccsm30"].interpolate(inplace=True)
df["nmme-prate-56w__ccsm3"].interpolate(inplace=True)
df["nmme-tmp2m-34w__ccsm3"].interpolate(inplace=True)
df["ccsm30"].interpolate(inplace=True)
```

```python
# round the float values to 2 decimal places to match the other values
df["nmme0-tmp2m-34w__ccsm30"] = df["nmme0-tmp2m-34w__ccsm30"].round(2)
df["nmme-tmp2m-56w__ccsm3"] = df["nmme-tmp2m-56w__ccsm3"].round(2)
df["nmme-prate-34w__ccsm3"] = df["nmme-prate-34w__ccsm3"].round(2)
df["nmme0-prate-56w__ccsm30"] = df["nmme0-prate-56w__ccsm30"].round(2)
df["nmme0-prate-34w__ccsm30"] = df["nmme0-prate-34w__ccsm30"].round(2)
df["nmme-prate-56w__ccsm3"] = df["nmme-prate-56w__ccsm3"].round(2)
df["nmme-tmp2m-34w__ccsm3"] = df["nmme-tmp2m-34w__ccsm3"].round(2)
df["ccsm30"] = df["ccsm30"].round(2)
```

```python
# checking if there are any null values left
null_dict = df.isnull().sum().to_dict()
null_dict
```

## 3- Statistical Summaries

```python
# observing different forecast models temperature means for different climate regions
df.groupby("climateregions__climateregion")[temp_group].mean().round(2)
```

```python
# observing different forecast models precipitation means for different climate regions
df.groupby("climateregions__climateregion")[prate_group].mean().round(2)
```

```python
# observing different forecast models temperature means for different months
df.groupby(df["startdate"].dt.month)[temp_group].mean().round(2)
```

```python
# observing different forecast models precipitation means for different years and climate regions
df.groupby([df["startdate"].dt.year, "climateregions__climateregion"])[
    prate_group
].mean().round(2)
```

```python
# observing different forecast models temperature means for different years and climate regions
df.pivot_table(
    index=df["startdate"].dt.year,
    columns="climateregions__climateregion",
    values=temp_group,
    aggfunc="mean",
).round(2)
```

```python
# observing different forecast models precipitation means for different years and climate regions
df.pivot_table(
    index=df["startdate"].dt.year,
    columns="climateregions__climateregion",
    values=prate_group,
    aggfunc="mean",
).round(2)
```

```python
# observing different forecast models temperature means for different months and locations
df.pivot_table(
    index=df[["lon", "lat"]],
    columns=df["startdate"].dt.month,
    values=temp_group,
    aggfunc="mean",
).round(2)
```

```python
# observing the relation between the mjo phase and the precipitation means
df.groupby("mjo1d__phase")[prate_group].mean()
```

```python
# observing the relation between the mjo amplitude and the precipitation means
df.groupby("mjo1d__amplitude")[prate_group].mean()
```

```python
# observing the relation between the Multivariate ENSO index and the temperature means
df.groupby("mei__mei")[temp_group].mean()
```

```python
# observing the relation between the Multivariate ENSO index rank and the temperature means
df.groupby("mei__meirank")[temp_group].mean()
```

## 4- Visualizing The Data

```python
plt.figure(figsize=(12, 6))
sns.histplot(x="nmme-tmp2m-34w__nmmemean", hue="climateregions__climateregion", data=df)
plt.title("Mean Temperature Histogram by Climate Region")
plt.xlabel("Mean Temperature (C°)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.kdeplot(x="nmme-tmp2m-34w__nmmemean", hue="climateregions__climateregion", data=df)
plt.title("Mean Temperature Distribution by Climate Region")
plt.xlabel("Mean Temperature (C°)")
plt.ylabel("Density")
plt.show()
```

```python
fig = sns.jointplot(
    x="nmme-prate-34w__nmmemean",
    y="nmme-tmp2m-34w__nmmemean",
    data=df,
    kind="hex",
    height=7,
)
plt.suptitle("Joint Distribution: Mean Precipitation vs Mean Temperature")
fig.set_axis_labels("Mean Precipitation (mm)", "Mean Temperature (C°)")
plt.tight_layout()
plt.show()
```

```python
pt = df.pivot_table(
    index=df["startdate"].dt.year,
    columns="mjo1d__phase",
    values="nmme-tmp2m-34w__nmmemean",
    aggfunc="mean",
).round(2)

plt.figure(figsize=(12, 6))
sns.heatmap(pt, annot=True, cmap="coolwarm")
plt.title("Mean Temperature Forecast by Year and MJO Phase")
plt.xlabel("MJO Phase")
plt.ylabel("Year")
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df, x="mei__nip", y="nmme-tmp2m-34w__nmmemean")

plt.title("Temperature Distribution by ENSO Phase")
plt.xlabel("Niño Index Phase")
plt.ylabel("Mean Temperature (C°)")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=df[0:731],
    x="mei__mei",
    y="nmme-tmp2m-34w__nmmemean",
    hue="mei__meirank",
    palette="coolwarm",
    alpha=0.8,
)

plt.title("Mean Temperature vs MEI (Colored by MEI Rank)")
plt.xlabel("Multivariate ENSO Index (MEI)")
plt.ylabel("Mean Temperature (C°)")
plt.legend(title="MEI Rank", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
fig = px.line(
    df[0:731],
    x="startdate",
    y="nmme-prate-34w__nmmemean",
    title="Rain Precipitation in The Dry Semi-arid steppe Hot (BSh) Climate Region",
)

fig.update_layout(xaxis_title="Date", yaxis_title="Mean Precipitation (mm)")
fig.show()
```

```python
fig = px.box(
    df,
    x="climateregions__climateregion",
    y="nmme-tmp2m-34w__nmmemean",
    title="Temperature Distribution by Climate Groups",
)

fig.update_layout(xaxis_title="Climate Region", yaxis_title="Mean Temperature (C°)")
fig.show()
```

```python
temp_date = temp_group
temp_date.append("startdate")
df_grouped = df.iloc[0:731][temp_date]
df_grouped = df_grouped[df_grouped["startdate"].dt.day == 15]
```

```python
fig = go.Figure()

for temp in temp_group:
    fig.add_trace(
        go.Scatter(
            x=df_grouped["startdate"],
            y=df_grouped[temp],
            mode="lines+markers",
            name=f"{temp}",
        )
    )

fig.update_layout(
    title="Temperature Forecast Models on the 15th Day of Each Month",
    xaxis_title="Date",
    yaxis_title="Mean Temperature (C°)",
    legend_title="Forecast Models",
    template="plotly_white",
)
fig.show()
```
