## Linear Algebra

Vectors and metrics

Vector: 1 row or 1 column

Metrics: multi dimentions nxm

```python
import numpy as np

# Define two sample matrices (metric tensors)
A = np.array([[1, 2],
              [3, 4]])
print('A = ', A, sep='\n')

B = np.array([[2, 1],
              [0, 3]])
print('B = ', B, sep='\n')

# Sum
sum_result = A + B

# Subtraction
subtraction_result = A - B

# Element-wise Multiplication
elementwise_multiplication_result = A * B

# Matrix Multiplication
matrix_multiplication_result = np.dot(A, B)

# Division (Multiplying A by inverse of B)
# Here, we'll assume B is invertible. If it's not, this operation would raise an error.
division_result = np.dot(A, np.linalg.inv(B))

print('A + B = ', sum_result, sep='\n', end="\n\n")
print('A - B = ', subtraction_result, sep='\n', end='\n\n')
print('A * B = ', elementwise_multiplication_result, sep='\n', end="\n\n")
print('A . B = ', matrix_multiplication_result, sep='\n', end="\n\n")
print('A / B = A . INV(B) = ', division_result, sep='\n', end="\n\n")
```

### Dot Product:

The dot product of two vectors is a scalar. It's calculated as the sum of the products of the corresponding components of the two vectors.
For two vectors
a.b

### Cross Product:

The cross product of two vectors results in another vector that is perpendicular to both of the original vectors.
It's defined only for three-dimensional vectors.
For vectors aXb


```python
# Define two sample vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Dot Product
dot_product_result = np.dot(a, b)

# Cross Product
cross_product_result = np.cross(a, b)

print("The result of the dot product: ",dot_product_result,end="\n\n")
print("The result of the cross product: ",cross_product_result,end="\n\n")
```

##Calculus

differentiation: Slope, Change rate, Gradient

integration: Area under the curve

```python
import sympy as sp

# Define the symbol
x = sp.symbols('x')

# Define the function
f = x**2 + 3*x + 2

# Differentiation
derivative_f = sp.diff(f, x)

# Indefinite Integration
indefinite_integral = sp.integrate(f, x)

# Definite Integration from 0 to 2
definite_integral = sp.integrate(f, (x, 0, 2))
```

```python
# plots for deep understanding

import matplotlib.pyplot as plt
import numpy as np

# Define the function and its derivative using numpy for plotting
x_vals = np.linspace(-5, 5, 400)
f_vals = x_vals**2 + 3*x_vals + 2
derivative_vals = 2*x_vals + 3

# Plotting the function
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x_vals, f_vals, label='f(x) = $x^2 + 3x + 2$', color='blue')
plt.title('Function: $f(x)$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

# Plotting the function with its derivative
plt.subplot(1, 3, 2)
plt.plot(x_vals, f_vals, label='f(x) = $x^2 + 3x + 2$', color='blue')
plt.plot(x_vals, derivative_vals, label="f'(x) = $2x + 3$", color='red', linestyle='--')
plt.title('Function & its Derivative')
plt.xlabel('x')
plt.grid(True)
plt.legend()

# Plotting the function with its area (from x=0 to x=2)
plt.subplot(1, 3, 3)
plt.plot(x_vals, f_vals, label='f(x) = $x^2 + 3x + 2$', color='blue')
plt.fill_between(x_vals, f_vals, where=[(0 <= xi <= 2) for xi in x_vals], color='gray', alpha=0.5, label='Area under curve')
plt.title('Function & Area under Curve')
plt.xlabel('x')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

#Statistics

what are the types of statistics?

1) Descriptive statistics

Measure central tendency (Mean, median, mode)

Measure of variability (Range, Variance and Standard deviation)

Measure of spread (5 number summaries, IQR, Outliers)

2) Inferential statistics

Studying a sample and generalizing results on the full population.

###Descriptive Statistics

i. Measure of Central Tendency:

Mean: The average of the numbers.

Median: The middle value when the numbers are sorted.

Mode: The number that occurs most frequently.

ii. Measure of Variability:

Range: The difference between the highest and lowest values.

Variance: The average of the squared differences from the mean.

Standard Deviation: The square root of the variance.

iii. Measure of Spread:


5 Number Summary:

This includes the minimum, first quartile (Q1), median (Q2), third quartile (Q3), and the maximum. The Interquartile Range (IQR) is the difference between Q3 and Q1 and represents the range within which the central 50% of the values lie.

iv. Outliers:

Typically, data points that fall below

َQ1-1.5*IQR

and above

َQ3+1.5*IQR

are considered outliers.


```python
# Sample data for demonstration
from scipy import stats

data = np.array([15, 20, 21, 22, 23, 23, 25, 26, 28, 29, 30, 35, 40, 45, 50, 55, 60, 100])

# i. Measure of Central Tendency
mean_data = np.mean(data)
median_data = np.median(data)
mode_data = stats.mode(data)[0]

# ii. Measure of Variability
range_data = np.ptp(data)
variance_data = np.var(data)
std_dev_data = np.std(data)

print("The mean of the data is : ", mean_data)
print("The median of the data is : ", median_data)
print("The mode of the data is : ", mode_data)
print("The range of the data is : ", range_data)
print("The variance of the data is : ", variance_data)
print("The standard deviation of the data is : ", std_dev_data)
```

```python
# Importing the required module
from scipy import stats

# Recomputing the necessary values using the stats module
mean_data = np.mean(data)
std_dev_data = np.std(data)

# Plotting data points and variance again
plt.figure(figsize=(8, 6))

# Plotting the data points
plt.scatter(data, [0] * len(data), color='blue', s=40, label='Data Points', zorder=5)

# Indicating variance
mean_line = plt.axvline(x=mean_data, color='green', linestyle='--', label='Mean')
upper_bound = mean_data + std_dev_data
lower_bound = mean_data - std_dev_data
plt.fill_between([lower_bound, upper_bound], -1,1,  color='gray', alpha=0.5, label='Variance (1 Std Dev)')

plt.title('Data Points with Variance Indication')
plt.xlabel('Value')
plt.yticks([])
plt.legend()
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```

```python
# iii. Measure of Spread
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1
five_num_summary = [data.min(), q1, median_data, q3, data.max()]

# iv. Outliers
outliers = data[(data < (q1 - 1.5 * iqr)) | (data > (q3 + 1.5 * iqr))]

# Plot
plt.figure(figsize=(8, 6))
plt.boxplot(data, vert=False, widths=0.7)
plt.scatter(data, [1] * len(data), color='blue', s=40, label='Data Points')
plt.scatter(outliers, [1] * len(outliers), color='red', s=100, zorder=5, label='Outliers')
plt.title('Boxplot showing 5-number summary and outliers')
plt.xlabel('Value')
plt.yticks([])
plt.legend()
plt.grid(axis='x')

central_tendency = (mean_data, median_data, mode_data)
variability = (range_data, variance_data, std_dev_data)
spread = five_num_summary
outliers = outliers

central_tendency, variability, spread, outliers
```

## Probability



### 1. **Basic Probability Concept**:

**Probability** is the measure of the likelihood that an event will occur. It is quantified as a number between 0 and 1, inclusive, where 0 indicates impossibility and 1 indicates certainty. The probability \( P(E) \) of an event \( E \) is defined as:


P(E) = Number of favorable outcomes/Total number of outcomes

### Example 1: Rolling a Die

When rolling a fair six-sided die, what is the probability of getting a 5?

**Solution**:
- There is 1 favorable outcome (getting a 5).
- There are 6 possible outcomes (1, 2, 3, 4, 5, or 6).
  
Thus, the probability \( P(5) \) is \( 1/6 \).

### 2. **Compound Events**:

When dealing with multiple events, we can have:
- **Independent Events**: Events that don't affect each other.
- **Dependent Events**: The outcome of one event affects the outcome of the second event.
- **Mutually Exclusive Events**: Events that cannot happen at the same time.

### Example 2: Drawing Cards

Suppose we draw two cards from a standard deck of 52 cards. What is the probability that the first card is an Ace and the second card is a King?

**Solution**:
- The probability that the first card is an Ace is \( {4/52} \) since there are 4 Aces in the deck.
- After drawing an Ace, there are only 51 cards left. Thus, the probability that the second card is a King is \( {4/51} \) (assuming we don't replace the card).

Since the two events are dependent, the combined probability is:
\[ P({Ace then King}) = P({Ace}) \times P({King after drawing an Ace}) \]

Let's calculate these probabilities.

For the card drawing example:

- Probability of drawing an Ace: \( P({Ace}) \approx 0.0769 \) or \( {4/52} \)
- Probability of drawing a King after drawing an Ace: \( P(\text{King after Ace}) \approx 0.0784 \) or \( {4/51} \)
- Combined probability of drawing an Ace first and then a King: \( P(\text{Ace then King}) \approx 0.00603 \)

### 3. **Complementary Events**:

The probability that Event \( E \) does not happen is denoted by \( P(E') \) and is called the complementary probability. It's calculated as:

\[ P(E') = 1 - P(E) \]



```python
# Scenario: Roll a fair six-sided die and then flip a fair coin.

# Sample spaces for the events
dice_outcomes = [1, 2, 3, 4, 5, 6]  # Possible outcomes when rolling a dice
coin_outcomes = ['Heads', 'Tails']  # Possible outcomes when flipping a coin

# 1. Probability of rolling a 4
# There's one favorable outcome (rolling a 4) out of six possible outcomes.
prob_roll_4 = dice_outcomes.count(4) / len(dice_outcomes)

# 2. Probability of getting a Heads on the coin flip
# There's one favorable outcome (getting Heads) out of two possible outcomes.
prob_heads = coin_outcomes.count('Heads') / len(coin_outcomes)

# 3. Probability of rolling a 4 and getting a Heads
# As rolling a dice and flipping a coin are independent events, we multiply their probabilities.
prob_4_and_heads = prob_roll_4 * prob_heads

# 4. Probability of NOT rolling a 4
# This is the complementary event of rolling a 4.
prob_not_4 = 1 - prob_roll_4

# 5. Probability of NOT getting a Heads
# This is the complementary event of getting a Heads.
prob_not_heads = 1 - prob_heads

prob_roll_4, prob_heads, prob_4_and_heads, prob_not_4, prob_not_heads
```
